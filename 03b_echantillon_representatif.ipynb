## **1. Méthodologie de constitution de l'échantillon d'entraînement**

On cherche d'abord à générer un echantillon de **5000 quadruplets (evenement, où, quand, qui)**.

Pour générer cet échantillon on crée  plusieurs **bases de sondage ** intermédiaires  :

- **base 1** pour les '**événements** (de nombreux exemples avec bcp de diversité)

- **base 2** pour les **lieux** : noms de villes, de rues, d'avenue, boulevards, monuments, commerces, lieux publics, lieu, de transports etc. Il existe peut être une base officielle (Insee ou autre)

- **base 3** pour les expressions désognant des **moments** : des jours de semaine, des dates, des horaires, écrits sous tous les formats possibles et en les combinant entre eux de toutes les façons possibles

- **base 4a** pour les **prénoms** existant en France : pourquoi pas utiliser une base connnue de prénoms avec leurs fréquences respectives en France (peut être que de telles bases issues de travaux statistiques sont disponibles sur des sites officiels ?)

- **base 4b** : même chose pour les **noms de famille**

- **base 4c** pour les expressions désignant des individus **hors prénoms** et noms.  Existe-t-il une sorte de catégorisation officielle, par exemple : monsieur, madame, individu, agent, plombier, curé

- **base 4** consrtuite à partir des bases 4a, 4b et 4c (par tirages par exemple, mais en faisant atrention a la cohénrece : Monsieur par exemple ne peut pas etre suivi par un prenom feminin). Cette base contiendra un ensemble d'expressions désignant un individu : par son prénom et/ou son nom tirés dans la base des noms et celle des prénoms, éventuellement attachés à une expression comme Monsieur, Madame etc. ; mais aussi par les autres types d'expression désignant les individus, tirés dans la base de sondage ad hoc. On peut ou pas combiner nom, prénom, autres.

On génère alors un échantillon de 5000 quadruplets (événement, où, quand, qui) par tirages indépendants dans chacune des bases 1 (pour la composante evenement), 2 (pour la compsante où) 3 (pour la compoqabte quand), 4 (pour la composante qui).

A partir de cet échantillon, on tire **5000 textes** qui serviront d'inputs pour la base d'entraînement. Le texte numéro i fait intervenir les composantes numéro i des bases correspondantes.

On fait attention à ce que les tournures de phrases associées à ces textes recouvrent une grande diversité afin d'éviter des problèmes de representivite.

Enfin, pour chaque texte input, on génère le résumé formaté eu format json qui correspond.

**Ces bases ont été générées par ChatGPT, en prenant un prompt très proche de ce descriptif.**

Dans un premier temps, il a été demandé de générer 1000 événements différents. L'hétérogénité des événements était générée de façon artificielle par ChatGPT et reposait sur des croisements :

## **6. Construction d'un échantillon de quadruplets (evenement, lieu, moment, individus)**

```python
liste_evenements = base_evenement["evenement"].tolist()
liste_lieux = base_lieu["lieu"].tolist()
liste_moments = base_moment["quand"].tolist()
liste_individus = base_individu["qui"].tolist()
```

```python
import random

# Tirage indépendant de 5000 quadruplets
quadruplets = [
    {
        "evenement": random.choice(liste_evenements),
        "lieu": random.choice(liste_lieux),
        "moment": random.choice(liste_moments),
        "individus": random.choice(liste_individus)
    }
    for _ in range(5000)
]
```

## **7. Génération de l'échantillon d'entraînement pour le fine-tuning**

A partir de l'échantillon de quadruplets, on génère avec ChatGPT des textes qui sont des formes rédigées des composantes evenement, lieu, moment et individus.

C'est une tache qui a dû être réalisée de façon incrémentale.
Spontanément, ChatGPT génèrait des phrases maladroites :

- les items vides, marqués comme NaN, donnait lieu à des formulations étranges comme *NaN s'est rendu au 28 rue des tulipes au moment NaN*
- il a donc fallu mentionner à ChatGPT le fait de ne pas tout évoquer les items manquants
- mais aussi lui demander de générer des phrases les plus naturelles qui soient, en évitant les formulations qui adhéraient trop aux items. En particulier, cela passait par éviter de paraphraser ces items, les reformuler avec d'autres mots
- les phrases générées avaient tendance également à suivre toutes un peu la même formulation (maladroite) du type "il a été évoqué que A" ou "des témoins ont rapporté B". Ceci a amené à inclure dans le prompt une directive du type *pas deux fois la même formulation ni tournure de phrase d'une ligne à l'autre du tableau"
- certains textes étaient rédigés sans article défini ou indéfini, quand ils mentionnaient des lieux, des moments ou des individus : là aussi, il a fallu inclure dans le prompt la nécessité de bien mettre les articiles pour rendre chaque phrase correcte.

## **8. Conclusion**

Ces explorations ont montré que l'approche présentée ici, qui consiste à générer des échantillons représentatifs de chaque item, n'est finalment pas optimale. Autant il était possible de générer des échantillons corrects pour chaque item, autant la génréation de textes à partir de ces items a posé des problèmes qui n'ont pu être résolus.

Le principal problème n'est pas dans la génération des échantillons eux-mêmes, mais davantage dans celle d'un grand nombre de textes, sous contrainte des quadruplets *(événement, lieu, momnent individus)*. ChatGPT peine à satisfaire ces contraintes et a tendance à construire des textes extrêmement artificiels ou à la tournure très maladroite, voire incorrecte. Par ailleurs, il procède de la façon suivante :

- génération d'une (ou d'un nombre limité) de phrase(s) générique(s) qui peut s'appliquer à tout quadruplet *(événement, lieu, momnent individus)*
- application de ces phrases à ce quadruplet

Cela conduit à des exemples maladroits et récurrents, tels que :

- Il a été rapporté que **Léa, Vincent** ont été témoins de **braquage d'une banque** au moment **20/07/2025, 22h15**, à **8, rue des lilas, Nantes**

- Il a été rapporté que **M. Dupond, son amie, un avocat** ont été témoins de **enlèvement dans la rue** au moment **dimanche soir, 21h30**, à **gare SNCF de Marseille**
