**Définition de la fonction de découpage d'un texte**

- **input :**
  - le nom du fichier du texte à découper
  - le nom du répertoire qui contient les blocs à enregistrer
- **output :** les blocs de texte, au format json, enregistrés dans le répertoire précisé en input

```python
import spacy
from sentence_transformers import SentenceTransformer, util

def decouper_texte(nom_fichier, repertoire_blocs):

  # Charger le modèle spaCy pour le français
  nlp = spacy.load("fr_core_news_sm")

  # Charger le modèle de similarité sémantique
  model = SentenceTransformer("paraphrase-multilingual-MiniLM-L12-v2")

  # Lire le texte
  with open(nom_fichier, encoding="utf-8") as f:
      texte = f.read()

  # Segmenter en phrases avec spaCy
  doc = nlp(texte)
  phrases = [sent.text.strip() for sent in doc.sents if sent.text.strip()]

  # Embeddings des phrases
  embeddings = model.encode(phrases, convert_to_tensor=True)

  # Paramètres
  seuil_similarite = 0.65  # plus élevé = blocs plus courts, plus rigoureux

  # Regroupement en blocs
  blocs = []
  bloc_courant = [phrases[0]]

  for i in range(1, len(phrases)):
      sim = util.pytorch_cos_sim(embeddings[i], embeddings[i - 1]).item()

      if sim >= seuil_similarite:
          bloc_courant.append(phrases[i])
      else:
          blocs.append(" ".join(bloc_courant))
          bloc_courant = [phrases[i]]

  # Ajouter le dernier bloc
  if bloc_courant:
      blocs.append(" ".join(bloc_courant))

  # Affichage ou export
  for i, bloc in enumerate(blocs):
      print(f"\n🧩 Bloc {i + 1} :\n{bloc}\n")
```

**La fonction *decouper_texte()* construit des blocs de façon cohérente.**

Dans l'exemple qui suit, on fournit à la fonction un document texte de 33 pages.
La fonction construit 1 480 blocs.

**temps d'exécution : 35 s**

```python
import time

start = time.time()

decouper_texte("recueil_nouvelles (1).txt", repertoire_blocs = "test")

end = time.time()
print(f"Temps d'exécution : {end - start:.2f} secondes")
```
